```python

# snake water gun game
import random
def game():
  print("Welcome to Snake-Water-Gun Game")
  print("Choices: \n1. Snake\n2. Water\n3. Gun")
  choices=['snake','water','gun']
  # user choice
  user_choice=input("Enter your choice(snake/water/gun).....").lower()
  if user_choice not in choices:
    print("Invalid choice! Pls select snake,water, or gun.")
    return
  # computer choice
  comp_choice=random.choice(choices)
  print(f"Cmputer choice :{comp_choice}")
  #game logic
  if user_choice==comp_choice:

    print("It's draw ")
  elif(user_choice=='snake' and comp_choice=='water') or (user_choice=='water' and comp_choice=='gun') or (user_choice=='gun' and comp_choice=='snake'):
    print("You win")
  else:
    print("Computer wins!")

game()

```
